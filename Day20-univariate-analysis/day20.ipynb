```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

# 1. Categorical Data
### a. Countplot

```python
sns.countplot(x = df['Survived'])
df['Survived'].value_counts()
```

```python
df['Survived'].value_counts().plot(kind='bar')
```

```python
# Here we are plotting the count of passengers in each class
sns.countplot(x='Pclass', hue='Pclass', data=df, palette=['#FF6347', '#4682B4', '#32CD32'], legend=False)


# x='Pclass': এটি নির্দেশ করে যে X-অক্ষে 'Pclass' কলামের মানগুলো দেখানো হবে।
# hue='Pclass': এটি নির্দেশ করে যে 'Pclass' কলামের প্রতিটি মানের জন্য আলাদা রঙ ব্যবহার করা হবে। এখানে hue এবং x একই ভ্যারিয়েবল ব্যবহার করছে।
# palette=['#FF6347', '#4682B4', '#32CD32']: এটি তিনটি আলাদা রঙ নির্দিষ্ট করে, যা 'Pclass'-এর তিনটি মান (1, 2, 3) এর জন্য ব্যবহৃত হবে।
# legend=False: এটি প্লটে লেজেন্ড বন্ধ করে দেয়, যাতে প্লটটি পরিষ্কার দেখায়।
```
